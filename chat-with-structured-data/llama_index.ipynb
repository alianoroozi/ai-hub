```python
import pandas as pd

DATA_FILE = "data/ml_interview_dataset.parquet"
df = pd.read_parquet(DATA_FILE)
df["delivery_date"] = pd.to_datetime(df["delivery_date"])
df["ordered_date"] = pd.to_datetime(df["ordered_date"])
```

```python
from llama_index.experimental.query_engine import PandasQueryEngine
```

```python
query_engine = PandasQueryEngine(df=df, verbose=True)
```

```python
user_query = "Show me the total cost of items ordered by the Electronics department from vendor OfficeMax Corp in feb 2024."
response = query_engine.query(user_query)
```

```python
query_engine = PandasQueryEngine(df=df, verbose=True, synthesize_response=True)
response = query_engine.query(user_query)
print(str(response))
```
